## 4. Quick prediction sanity check

````python
import joblib
from common.features import FEATURE_COLUMNS, to_vector
bundle = joblib.load('models/model.pkl')
print('classes:', bundle['classes'])
# a 'suspicious-looking' row: high file rate + powershell + spawn burst
row = {c: 0 for c in FEATURE_COLUMNS}
row.update(file_event_rate=40, suspicious_proc_count=4, new_process_count=9, cpu_pct=60)
print('P(classes) =', bundle['rf'].predict_proba([to_vector(row)])[0])
````
